### Classify Leaks

```python
def classify_leak(row):
    if row["Profit"] >= 0:
        return "No Leak"
    if row["Discount"] > 0.40:
        return "Extreme Discount"
    if row["Discount"] > 0.20:
        return "High Discount"
    if row["Profit Margin"] < -20:
        return "Severe Margin Loss"
    return "Low Margin Loss"
df["Leak Type"] = df.apply(classify_leak, axis=1)
```

```python
leak_type_analysis = (
    df.groupby("Leak Type")
      .agg(
          Records=("Profit", "count"),
          Revenue=("Sales", "sum"),
          Profit=("Profit", "sum")))

leak_type_analysis["Profit Margin"] = (
    leak_type_analysis["Profit"] /
    leak_type_analysis["Revenue"]) * 100

leak_type_analysis.sort_values("Profit")
```

### Where Should a Business Investigate First 
#### we will combine:
- Number of Loss Records
- Total Financial loss
- Loss Margin

```python
leak_impact = (
    df[df["Profit"] < 0]
    .groupby("Sub-Category")
    .agg(
        Loss_Records=("Profit", "count"),
        Total_Loss=("Profit", "sum"),
        Revenue=("Sales", "sum")
    )
)

leak_impact["Loss_Per_Record"] = (
    abs(leak_impact["Total_Loss"]) /
    leak_impact["Loss_Records"]
)

leak_impact["Loss_Margin"] = (
    leak_impact["Total_Loss"] /
    leak_impact["Revenue"]
) * 100

leak_impact.sort_values(
    "Total_Loss"
)
```

## Build Profit Leak Score
- Total loss → overall financial impact
- Loss records → frequency
- Loss per record → severity

```python
leak_impact["Loss_Share"] = (
    abs(leak_impact["Total_Loss"]) /
    abs(leak_impact["Total_Loss"]).sum()
) * 100

leak_impact
```

### Step 6 — Create a ProfitGuard Priority Score

We'll normalize the three measures so they're comparable:

- Loss Share → financial impact
- Loss Records → frequency
- Loss Per Record → severity

```python
leak_impact["Impact_Score"] = (
    leak_impact["Loss_Share"] /
    leak_impact["Loss_Share"].max()
) * 100

leak_impact["Frequency_Score"] = (
    leak_impact["Loss_Records"] /
    leak_impact["Loss_Records"].max()
) * 100

leak_impact["Severity_Score"] = (
    leak_impact["Loss_Per_Record"] /
    leak_impact["Loss_Per_Record"].max()
) * 100

leak_impact
```

### Create he Priority Score

```python
leak_impact["Priority_Score"] = (
    0.50 * leak_impact["Impact_Score"] +
    0.30 * leak_impact["Frequency_Score"] +
    0.20 * leak_impact["Severity_Score"]
)

leak_impact = leak_impact.sort_values(
    "Priority_Score",
    ascending=False
)

leak_impact
```

## Anomaly Detection

```python
from sklearn.ensemble import IsolationForest

ml_features = df[
    ["Sales", "Quantity", "Discount", "Profit", "Profit Margin"]
].copy()

ml_features = ml_features.replace(
    [np.inf, -np.inf],
    np.nan
).dropna()

print("Rows available for ML:", len(ml_features))
ml_features.head()
```

```python
model = IsolationForest(
    contamination=0.05,
    random_state=42
)

model.fit(ml_features)

df.loc[ml_features.index, "Anomaly"] = model.predict(ml_features)

df["Anomaly"].value_counts()
```

## anomaly analysis

```python
anomaly_analysis = (
    df.groupby("Anomaly")
      .agg(
          Records=("Profit", "count"),
          Revenue=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Loss_Records=("Profit", lambda x: (x < 0).sum())
      )
)

anomaly_analysis["Profit Margin"] = (
    anomaly_analysis["Profit"] /
    anomaly_analysis["Revenue"]
) * 100

anomaly_analysis["Loss Rate"] = (
    anomaly_analysis["Loss_Records"] /
    anomaly_analysis["Records"]
) * 100

anomaly_analysis
```

### The important finding is the Loss Rate:

- Anomalies → 65.73%
- Normal → 16.26%

## ML-backed ProfitGuard signal

```python
df["ML_Profit_Leak"] = np.where(
    (df["Anomaly"] == -1) &
    (df["Profit"] < 0),
    "Yes",
    "No"
)

ml_leaks = df[df["ML_Profit_Leak"] == "Yes"]

print("ML Profit Leak Candidates:", len(ml_leaks))
print("Total Loss:", ml_leaks["Profit"].sum())
```

### Instead of our earlier definition:

Profit < 0 → Profit Leak

we now have:

Unusual transaction + negative profit → ML Profit Leak Candidate

<p> ProfitGuard uses unsupervised anomaly detection to identify unusual transactions that are disproportionately associated with negative profitability, creating a focused investigation queue rather than flagging every loss-making transaction.
</p>

## Give every ML leak an explanation

```python
def explain_leak(row):
    reasons = []

    if row["Profit"] < 0:
        reasons.append("Negative Profit")

    if row["Discount"] > 0.40:
        reasons.append("Extreme Discount")
    elif row["Discount"] > 0.20:
        reasons.append("High Discount")

    if row["Profit Margin"] < -20:
        reasons.append("Severe Margin Loss")

    if row["Anomaly"] == -1:
        reasons.append("Unusual Transaction Pattern")

    return " + ".join(reasons)


df["Leak Reason"] = df.apply(explain_leak, axis=1)

ml_leaks = df[df["ML_Profit_Leak"] == "Yes"].copy()

ml_leaks[
    [
        "Sub-Category",
        "Sales",
        "Quantity",
        "Discount",
        "Profit",
        "Profit Margin",
        "Leak Reason"
    ]
].head(10)
```

## final ProfitGuard flags

```python
df["Negative_Profit"] = df["Profit"] < 0

df["High_Discount"] = df["Discount"] > 0.20

df["Severe_Margin"] = df["Profit Margin"] < -20

df["ML_Anomaly"] = df["Anomaly"] == -1

df["Leak_Candidate"] = (
    df["Negative_Profit"] &
    df["ML_Anomaly"]
)

df[
    [
        "Sub-Category",
        "Profit",
        "Discount",
        "Profit Margin",
        "Negative_Profit",
        "High_Discount",
        "Severe_Margin",
        "ML_Anomaly",
        "Leak_Candidate"
    ]
].head(10)
```

## Evidence Score 

```python
df["Evidence_Count"] = (
    df["Negative_Profit"].astype(int) +
    df["High_Discount"].astype(int) +
    df["Severe_Margin"].astype(int) +
    df["ML_Anomaly"].astype(int)
)

df[
    [
        "Sub-Category",
        "Profit",
        "Discount",
        "Profit Margin",
        "Evidence_Count",
        "Leak_Candidate",
        "Leak Reason"
    ]
].sort_values(
    "Evidence_Count",
    ascending=False
).head(20)
```

<p>Why this is useful

A transaction with:

- Evidence Count = 4

<u>has all four warning signals.</u>

A transaction with:

- Evidence Count = 2

<u>has two independent signals.</u>

And:

- Evidence Count = 0

<u>has none.</u>

This gives us an explainable evidence-strength measure, rather than pretending the number is a statistically calibrated probability.</p>

##  Now let's find the business root cause

<b>We've already established:</b>

Transaction → ML anomaly → candidate

Region → State → geographic concentration

Now we need to answer:

What products are actually driving these losses?

```python
category_leaks = (
    profit_leaks
    .groupby(["Category", "Sub-Category"])
    .agg(
        Leak_Records=("Profit", "count"),
        Revenue=("Sales", "sum"),
        Total_Loss=("Profit", "sum")
    )
)

category_leaks["Loss_Share"] = (
    abs(category_leaks["Total_Loss"]) /
    abs(category_leaks["Total_Loss"]).sum()
) * 100

category_leaks = category_leaks.sort_values(
    "Total_Loss"
)

category_leaks
```

```python
## Discount Leaks

discount_leaks = (
    profit_leaks
    .groupby("Discount Band")
    .agg(
        Leak_Records=("Profit", "count"),
        Revenue=("Sales", "sum"),
        Total_Loss=("Profit", "sum")
    )
)

discount_leaks["Loss_Margin"] = (
    discount_leaks["Total_Loss"] /
    discount_leaks["Revenue"]
) * 100

discount_leaks["Loss_Share"] = (
    abs(discount_leaks["Total_Loss"]) /
    abs(discount_leaks["Total_Loss"]).sum()
) * 100

discount_leaks
```

## Which products are losing money specifically under extreme discounting?

```python
root_cause = (
    profit_leaks
    .groupby(["Sub-Category", "Discount Band"])
    .agg(
        Leak_Records=("Profit", "count"),
        Revenue=("Sales", "sum"),
        Total_Loss=("Profit", "sum")
    )
)

root_cause["Loss_Share"] = (
    abs(root_cause["Total_Loss"]) /
    abs(root_cause["Total_Loss"]).sum()
) * 100

root_cause = root_cause.sort_values(
    "Total_Loss"
)

root_cause.head(20)
```

#  <b><u>Business Insight</u></b>

```python
root_cause["Loss_Per_Record"] = (
    abs(root_cause["Total_Loss"]) /
    root_cause["Leak_Records"]
)

root_cause = root_cause.sort_values(
    "Loss_Per_Record",
    ascending=False
)

root_cause.head(15)
```

## Final  Root-cause priority score

```python
root_cause["Impact_Score"] = (
    root_cause["Loss_Share"] /
    root_cause["Loss_Share"].max()
) * 100

root_cause["Frequency_Score"] = (
    root_cause["Leak_Records"] /
    root_cause["Leak_Records"].max()
) * 100

root_cause["Severity_Score"] = (
    root_cause["Loss_Per_Record"] /
    root_cause["Loss_Per_Record"].max()
) * 100

root_cause["Priority_Score"] = (
    0.50 * root_cause["Impact_Score"] +
    0.30 * root_cause["Frequency_Score"] +
    0.20 * root_cause["Severity_Score"]
)

root_cause = root_cause.sort_values(
    "Priority_Score",
    ascending=False
)

root_cause.head(15)
```
